```python
import numpy as np
import matplotlib.pyplot as plt
d=50 
A=np.loadtxt('0.4_stress.dat',skiprows=1)
A[:,0]=A[:,0]*5 #timestep=5
first_five = A[:5, :]
rest = A[5:A.shape[0]:d, :] #Note: I assume we keep first 5
A= np.vstack((first_five, rest))

A1=np.loadtxt('0.6_stress.dat',skiprows=1)
A1[:,0]=A1[:,0]*5 
first_five = A1[:5, :]
rest = A1[5:A1.shape[0]:d, :]
A1= np.vstack((first_five, rest))

A2=np.loadtxt('0.8_stress.dat',skiprows=1)
A2[:,0]=A2[:,0]*5 
first_five = A2[:5, :]
rest = A2[5:A2.shape[0]:d, :]
A2= np.vstack((first_five, rest))

difn=10 #note that 10 is tuned 
Zfit1=[]
Zfit2=[]
Zfit3=[]
z=[]
for i in range(A.shape[0]-difn):
    a=np.abs((((A[i,1]-A[i+difn,1])/(A[i,0]-A[i+difn,0]))*1000000000000000)/100000)  
    z=np.append(z,a)
plt.scatter(A[:z.shape[0],0]*0.000001,z,color='black',label='_nolegend_')
plt.plot(A[:z.shape[0],0]*0.000001,z,color='black')
Zfit1=np.append(Zfit1,z)
z0=np.copy(z)
z=[]
for i in range(A1.shape[0]-difn):
    a=np.abs((((A1[i,1]-A1[i+difn,1])/(A1[i,0]-A1[i+difn,0]))*1000000000000000)/100000)
    z=np.append(z,a)
plt.scatter(A1[:z.shape[0],0]*0.000001,z,color='red',label='_nolegend_')
plt.plot(A1[:z.shape[0],0]*0.000001,z,color='red')
Zfit2=np.append(Zfit2,z)
z1=np.copy(z)
z=[]
for i in range(A2.shape[0]-difn):
    a=np.abs((((A2[i,1]-A2[i+difn,1])/(A2[i,0]-A2[i+difn,0]))*1000000000000000)/100000)
    z=np.append(z,a)
plt.scatter(A2[:z.shape[0],0]*0.000001,z,color='blue',label='_nolegend_')
plt.plot(A2[:z.shape[0],0]*0.000001,z,color='blue')
Zfit3=np.append(Zfit3,z)
z2=np.copy(z)
plt.xlim([-0.02,1.01]) #1.01
plt.ylim([-100,1600]) #1600
plt.xlabel('Time (ns)',fontsize=20, fontname='Times New Roman')
plt.ylabel('Strain rate (10\u2075×1/s)',fontsize=20, fontname='Times New Roman')
plt.legend([r'$0.4\sigma_y$',r'$0.6\sigma_y$',r'$0.8\sigma_y$'],fontsize=12.5)
plt.xticks(fontsize=20, fontname='Times New Roman')
plt.yticks(fontsize=20, fontname='Times New Roman') 
plt.tight_layout()
#plt.savefig('S9f.png', dpi=600)
```
