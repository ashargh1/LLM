```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

stress_strain=np.loadtxt('W-32.dat',skiprows=1)
strain = stress_strain[:, 1]
stress = stress_strain[:, 2]
elastic_region = slice(0, 20 )  # Fit on the first 21 points
linear_fit = np.polyfit(strain[elastic_region], stress[elastic_region], 1)
slope, intercept = linear_fit
offset_strain = 0.002 #0.2% offset line (y = mx + c)
offset_intercept = -slope * offset_strain
offset_line = slope * strain + offset_intercept
diff = stress - offset_line
plt.plot(strain, offset_line, '--', label='0.2% Offset Line')
plt.plot(strain, stress, label='Stress-Strain Curve')
plt.xlabel('Strain')
plt.ylabel('Stress (GPa)')
plt.ylim([0,1])
plt.xlim([0,0.2])
plt.grid(which='major', linestyle='-', linewidth=0.8)
plt.grid(which='minor', linestyle=':', linewidth=0.5)
plt.minorticks_on()  # Enable minor ticks
plt.show()
```
